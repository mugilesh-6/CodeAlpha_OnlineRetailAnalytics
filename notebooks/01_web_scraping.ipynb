```python
import sys
import os

# Add src directory to path
sys.path.append(os.path.join('..', 'src'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scraper import BooksScraper

# Set display options
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', 50)

# Configure plotting
plt.style.use('default')
sns.set_palette("husl")

print("Libraries imported successfully!")
```

## Initialize Web Scraper

The BooksScraper class handles:
- HTTP requests with proper headers
- HTML parsing using BeautifulSoup
- Rate limiting to be respectful to the server
- Error handling for robust scraping

```python
# Initialize the scraper
scraper = BooksScraper(
    base_url="https://books.toscrape.com",
    delay=1.0  # 1 second delay between requests
)

print("Scraper initialized successfully!")
print(f"Base URL: {scraper.base_url}")
print(f"Request delay: {scraper.delay} seconds")
```

## Test Scraping Single Page

Let's first test our scraper on a single page to verify it works correctly.

```python
# Test scraping first page
test_url = "https://books.toscrape.com/index.html"
test_books = scraper.scrape_books_from_page(test_url)

print(f"Books found on first page: {len(test_books)}")

if test_books:
    print("\nSample book data:")
    for key, value in list(test_books[0].items())[:5]:
        print(f"{key}: {value}")
```

## Full Website Scraping

Now let's scrape multiple pages. We'll limit to 5 pages for demonstration purposes.

```python
# Scrape books from multiple pages
print("Starting web scraping process...\n")

# Scrape up to 5 pages (about 100 books)
books_df = scraper.scrape_all_books(max_pages=5)

print(f"\nScraping completed!")
print(f"Total books scraped: {len(books_df)}")
```

## Data Exploration

Let's examine the scraped data structure and quality.

```python
# Display basic information about the dataset
print("DATASET OVERVIEW")
print("="*50)
print(f"Shape: {books_df.shape}")
print(f"Columns: {list(books_df.columns)}")
print("\nData Types:")
print(books_df.dtypes)
print("\nMissing Values:")
print(books_df.isnull().sum())
```

```python
# Display first few rows
print("SAMPLE DATA")
print("="*50)
books_df.head()
```

## Data Quality Analysis

```python
# Analyze price data
if 'Price' in books_df.columns:
    print("PRICE ANALYSIS")
    print("="*30)
    print(f"Average price: £{books_df['Price'].mean():.2f}")
    print(f"Median price: £{books_df['Price'].median():.2f}")
    print(f"Price range: £{books_df['Price'].min():.2f} - £{books_df['Price'].max():.2f}")
    print(f"Standard deviation: £{books_df['Price'].std():.2f}")

# Analyze ratings
if 'Rating' in books_df.columns:
    print("\nRATING ANALYSIS")
    print("="*30)
    print(f"Average rating: {books_df['Rating'].mean():.2f}")
    print(f"Rating distribution:")
    print(books_df['Rating'].value_counts().sort_index())
```

## Visualizations of Scraped Data

```python
# Create visualizations
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.suptitle('Scraped Books Data Analysis', fontsize=16, fontweight='bold')

# Price distribution
if 'Price' in books_df.columns:
    axes[0, 0].hist(books_df['Price'], bins=20, alpha=0.7, color='skyblue', edgecolor='black')
    axes[0, 0].set_title('Price Distribution')
    axes[0, 0].set_xlabel('Price (£)')
    axes[0, 0].set_ylabel('Frequency')
    axes[0, 0].grid(True, alpha=0.3)

# Rating distribution
if 'Rating' in books_df.columns:
    rating_counts = books_df['Rating'].value_counts().sort_index()
    axes[0, 1].bar(rating_counts.index, rating_counts.values, alpha=0.7, color='lightcoral')
    axes[0, 1].set_title('Rating Distribution')
    axes[0, 1].set_xlabel('Rating (Stars)')
    axes[0, 1].set_ylabel('Number of Books')
    axes[0, 1].grid(True, alpha=0.3)

# Price vs Rating scatter plot
if 'Price' in books_df.columns and 'Rating' in books_df.columns:
    scatter = axes[1, 0].scatter(books_df['Rating'], books_df['Price'], 
                                alpha=0.6, c=books_df['Price'], cmap='viridis')
    axes[1, 0].set_title('Price vs Rating')
    axes[1, 0].set_xlabel('Rating (Stars)')
    axes[1, 0].set_ylabel('Price (£)')
    axes[1, 0].grid(True, alpha=0.3)
    plt.colorbar(scatter, ax=axes[1, 0], label='Price (£)')

# Availability status
if 'Availability' in books_df.columns:
    availability_counts = books_df['Availability'].value_counts()
    axes[1, 1].pie(availability_counts.values, labels=availability_counts.index, 
                   autopct='%1.1f%%', startangle=90)
    axes[1, 1].set_title('Availability Status')

plt.tight_layout()
plt.show()
```

## Save Scraped Data

```python
# Save the scraped data
output_file = '../data/scraped/books_scraped.csv'

# Create directory if it doesn't exist
os.makedirs(os.path.dirname(output_file), exist_ok=True)

# Save to CSV
books_df.to_csv(output_file, index=False, encoding='utf-8')

print(f"Data successfully saved to: {output_file}")
print(f"File size: {os.path.getsize(output_file) / 1024:.1f} KB")
print(f"Records saved: {len(books_df)}")
```
